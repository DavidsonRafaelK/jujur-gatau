## Ekstraksi Informasi dan Metadata Sequence FASTA

function `get_fasta_info()` untuk ngebaca file FASTA dan meng-ekstrak 2 jenis informasi sekaligus:
- jumlah A, T, G, C, panjang sequence dan GC content
- metadata sequence (id, nama, description dan urutan sequence)

terus dipanggil di cell 5 buat baca 2 file beda dan ditampilin hasilnya buat ngecek karakteristik masing2 sequence.

```python
def get_fasta_info(fasta_file):
  record = SeqIO.read(fasta_file, "fasta") # baca file fasta
  sequence = str(record.seq).upper()       # ubah jadi string huruf kapital

  jumlahG = sequence.count('G')
  jumlahC = sequence.count('C')
  jumlahA = sequence.count('A')
  jumlahT = sequence.count('T')

  gc_content = ((jumlahG + jumlahC) / (jumlahA + jumlahT + jumlahG + jumlahC)) * 100

  return {
      "id": record.id,
      "panjang": len(sequence),
      "G": jumlahG,
      "C": jumlahC,
      "A": jumlahA,
      "T": jumlahT,
      "GC_content": round(gc_content, 2),
      "metadata_title": record.name,
      "metadata_desc": record.description,
      "metadata_seq": record.seq,
      "metadata_id": record.id,
  }
```

```python
data1 = get_fasta_info("./JBFSEQ010000003.1.fasta")
data2 = get_fasta_info("./NM_007294.4.fasta")

print("Data 1 (JBFSEQ010000003.1):")
print(data1['id'])
print(data1['panjang'])
print(data1['G'])
print(data1['C'])
print(data1['A'])
print(data1['T'])
print(data1['GC_content'])
print(data1['metadata_title'])
print(data1['metadata_desc'])
print(str(data1['metadata_seq'])[:50]) # ambil 50 baris pertama aja
print(data1['metadata_id'])

print()

print("Data 2 (NM_007294.4):")
print(data2['id'])
print(data2['panjang'])
print(data2['G'])
print(data2['C'])
print(data2['A'])
print(data2['T'])
print(data2['GC_content'])
print(data2['GC_content'])
print(data2['metadata_title'])
print(data2['metadata_desc'])
print(str(data2['metadata_seq'])[:50]) # ambil 50 baris pertama aja
print(data2['metadata_id'])
```

## Motif search FASTA

```python
def find_motif_in_fasta(fasta_file, motif):
    record = SeqIO.read(fasta_file, "fasta")  # baca file fasta
    sequence = str(record.seq).upper()        # ubah jadi string huruf kapital
    motif = motif.upper()                     # samakan case dengan sequence

    posisi = sequence.find(motif)

    if posisi != -1:
        ditemukan = True
        pesan = f"Motif '{motif}' ditemukan pada posisi {posisi} (0-based)"
    else:
        ditemukan = False
        pesan = f"Motif '{motif}' tidak ditemukan"

    return {
        "id": record.id,
        "motif": motif,
        "ditemukan": ditemukan,
        "posisi": posisi,      # -1 kalau tidak ketemu
        "pesan": pesan,
    }
```

```python
print(find_motif_in_fasta("./JBFSEQ010000003.1.fasta", "GGCC")['id'])
print(find_motif_in_fasta("./JBFSEQ010000003.1.fasta", "GGCC")['motif'])
print(find_motif_in_fasta("./JBFSEQ010000003.1.fasta", "GGCC")['ditemukan'])
print(find_motif_in_fasta("./JBFSEQ010000003.1.fasta", "GGCC")['posisi'])
print(find_motif_in_fasta("./JBFSEQ010000003.1.fasta", "GGCC")['pesan'])

print()

print(find_motif_in_fasta("./NM_007294.4.fasta", "GGCC")['id'])
print(find_motif_in_fasta("./NM_007294.4.fasta", "GGCC")['motif'])
print(find_motif_in_fasta("./NM_007294.4.fasta", "GGCC")['ditemukan'])
print(find_motif_in_fasta("./NM_007294.4.fasta", "GGCC")['posisi'])
print(find_motif_in_fasta("./NM_007294.4.fasta", "GGCC")['pesan'])
```

## Mentranslate codon berdasarkan tabel IUPAC

```python
def translate_codon(cod):
    """Translates a codon into an aminoacid using an internal
    dictionary with the standard genetic code."""
    tc = {
        "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
        "TGT": "C", "TGC": "C",
        "GAT": "D", "GAC": "D",
        "GAA": "E", "GAG": "E",
        "TTT": "F", "TTC": "F",
        "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G",
        "CAT": "H", "CAC": "H",
        "ATA": "I", "ATT": "I", "ATC": "I",
        "AAA": "K", "AAG": "K",
        "TTA": "L", "TTG": "L", "CTT": "L", "CTC": "L", "CTA": "L", "CTG": "L",
        "ATG": "M", "AAT": "N", "AAC": "N",
        "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
        "CAA": "Q", "CAG": "Q",
        "CGT": "R", "CGC": "R", "CGA": "R", "CGG": "R", "AGA": "R", "AGG": "R",
        "TCT": "S", "TCC": "S", "TCA": "S", "TCG": "S", "AGT": "S", "AGC": "S",
        "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
        "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
        "TGG": "W",
        "TAT": "Y", "TAC": "Y",
        "TAA": "_", "TAG": "_", "TGA": "_"
    }

    if cod in tc:
        return tc[cod]
    else:
        return None
```

```python
files = {
    "JBFSEQ010000003.1": "./JBFSEQ010000003.1.fasta",
    "NM_007294.4": "./NM_007294.4.fasta"
}

translated_sequences = {}

for name, filepath in files.items():
    seq = get_fasta_info(filepath)['metadata_seq']

    translated_seq = ""
    for i in range(0, min(len(seq), 45), 3):
        codon = seq[i:i+3]
        if len(codon) == 3:
            amino_acid = translate_codon(codon)
            translated_seq += amino_acid if amino_acid is not None else 'X'

    translated_sequences[name] = translated_seq
    print(f"Translated sequence for {name} (first 15 codons): {translated_seq}")
```
